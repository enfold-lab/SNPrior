### Figure 2. Accuracy by n_select and selection method

```r
df_for_plot = df %>% 
  filter(train_model == "SVM", 
         pre_select_method == "fst", #variance, random
         n_pre_select == 1000000,
         # n_dim_reduced == 1024,
         !(select_method %in% c("fst", "af")),
         select_n == n_dim_reduced
        )


p = ggplot(data = df_for_plot, aes(x = select_n, y = testset_accuracy, group = plot_unit, colour = select_method)) +
  geom_line(linewidth = 1) +
  xlab("# of features") + 
  ylab("Testset Accuracy (%)") +
  scale_color_manual(values = color_map_use) +
  scale_x_continuous(trans='log2') +
  #scale_y_continuous(limits = c(0,1)) +
  theme_bw() + 
  theme(panel.border = element_blank(), panel.grid.major = element_blank(),
        panel.grid.minor = element_blank(), axis.line = element_line(colour = "black")); 
        print(p)
```

```r
df_for_plot_summary = df_for_plot %>%
  group_by(pre_select_method, n_pre_select,	n_pre_select_goal, select_n, select_method) %>%
  mutate(target_val = testset_accuracy) %>%
  # mutate(target_val = testset_f1_macro) %>%
  summarise(
    sd = sd(target_val),
    sem = sd(target_val)/ sqrt(n()),
    target_val = mean(target_val),
    n = n()
  )
df_for_plot_summary %>% head()


p = ggplot(data = df_for_plot, aes(x = select_n, y = target_val, colour = select_method)) +
  # geom_jitter(position = position_jitter(0.2)) + 
  geom_line(aes(group = select_method),data = df_for_plot_summary) +
  geom_errorbar(aes(ymin = target_val-sem, ymax = target_val+sem), data = df_for_plot_summary, width = 0.2)+
  ggtitle("With pre-selection (Fst)") +
  xlab("# of features selected") + 
  ylab("Accuracy") +
  # ylab("Macro F1 score") +
  scale_color_manual(values = color_map_use) +
  scale_x_continuous(trans='log2', breaks = c(100, 1000, 10000, 100000)) +
  scale_y_continuous(limits = c(0.2,1)) +
  theme_bw() + 
  theme(panel.border = element_blank(), panel.grid.major = element_blank(),
        panel.grid.minor = element_blank(), axis.line = element_line(colour = "black")); 
print(p)

ggsave("results/fig2_accuracy_by_select_n.pdf", width=12, height=9, units = c("cm"), dpi = 300)
```

# Figure 3b. Accuracy by classes

```r
df_for_plot = df %>% 
  filter(train_model == "SVM_fixed", 
         pre_select_method == "random",
         n_pre_select == 1000000,
         select_method == "random",
         select_n == 1000000,
         select_n == n_dim_reduced
        ) %>%
  pivot_longer(
    cols = starts_with("testset_class_") & ends_with("_accuracy"),
    names_to = "class",
    values_to = "testset_accuracy_cls"
  ) %>%
  mutate(
    class = sub("^testset_class_", "", class),
    class = sub("_accuracy$", "", class)
  ) %>%
  select(random_seed, random_seed_cv, outer_fold, select_n, select_method, class, testset_accuracy_cls)
```

```r
df_for_plot_summary = df_for_plot %>%
  group_by(class) %>%
  mutate(target_val = testset_accuracy_cls) %>%
  # mutate(target_val = testset_f1_macro) %>%
  summarise(
    sd = sd(target_val),
    sem = sd(target_val)/ sqrt(n()),
    target_val = mean(target_val),
    n = n()
  ) %>% 
  arrange(desc(target_val)) %>%
  mutate(class = factor(class, levels = class))

df_for_plot_summary


p = ggplot(df_for_plot_summary, aes(x = class, y = target_val)) +
  geom_col(width = 0.75, fill = color_map_use[2]) +
  geom_errorbar(aes(ymin = target_val - sem, ymax = target_val + sem), width = 0.2, linewidth = 0.5) +
  coord_cartesian(ylim = c(0, 1)) +
  labs(x = NULL, y = "Test-set accuracy") +
  theme_classic(base_size = 10) +
  theme(axis.text.x = element_text(angle = 45, hjust = 1, vjust = 1), axis.ticks.x = element_blank())

p
print(p)

ggsave("results/fig3b_accuracy_by_cls.pdf", width=12, height=9, units = c("cm"), dpi = 300)
```

```r
df_line_plot = bind_rows(
    df_for_plot %>% transmute(class_target, select_n, dataset = "Validation", f1_macro = valset_f1_macro),
    df_for_plot %>% transmute(class_target, select_n, dataset = "Test", f1_macro = testset_f1_macro)
)

ggplot(df_line_plot, aes(x = select_n, y = f1_macro, color = dataset, group = dataset)) +
    geom_line(linewidth = 0.7) +
    geom_point(size = 1.5) +
    geom_hline(yintercept = 0.5, linetype = "dashed") +
    scale_x_log10() +
    scale_y_continuous(limits = c(0.4, 1)) +
    scale_color_manual(values = color_map_use[2:3]) +
    facet_wrap(~ class_target, ncol = 5) +
    labs(x = "Number of selected SNPs", y = "Macro F1", color = NULL) +
    theme_classic() +
    theme(legend.position = "top", strip.background = element_blank())
```

```r
f1_tolerance = 0.02

df_minimum_n_by_cls = df_for_plot %>% 
    mutate(max_valset_f1 = max(valset_f1_macro), 
           max_valset_f1_n = select_n[which.max(valset_f1_macro)]) %>%
    filter(valset_f1_macro >= (max_valset_f1 - f1_tolerance)) %>%
    slice_min(select_n, n = 1, with_ties = FALSE) 

df_minimum_n_by_cls    
    
df_minimum_n_by_cls %>% write.table("results/minimum_n_by_cls.csv", sep = ",")

## Continue plots in geography_plot.ipynb
```

```r
df_all_snps_by_cls = df_for_plot %>%
    filter(select_n == max(select_n))

df_testset_comparison <- bind_rows(
    df_minimum_n_by_cls %>% transmute(class_target, model = "Minimum SNPs", select_n, testset_f1_macro),
    df_all_snps_by_cls %>% transmute(class_target, model = "All SNPs", select_n, testset_f1_macro)
)  %>%
    mutate(model = factor(model, levels = c("Minimum SNPs", "All SNPs")))

class_order = df_testset_comparison %>%
    filter(model == "Minimum SNPs") %>%
    arrange(desc(testset_f1_macro)) %>%
    pull(class_target)

df_testset_comparison = df_testset_comparison %>%
    mutate(class_target = factor(class_target, levels = class_order))


p = ggplot(df_testset_comparison, aes(x = class_target, y = testset_f1_macro, fill = model)) +
    geom_col(position = position_dodge(width = 0.8), width = 0.7) +
    # geom_text(aes(label = select_n), position = position_dodge(width = 0.8), angle = 90, hjust = -0.15, size = 2.5) +
    scale_fill_manual(values = color_map_use[c(4, 2)]) +
    scale_y_continuous(limits = c(0, 1.02), breaks = seq(0, 1, 0.25), expand = c(0, 0)) +
    ggtitle("Minimum SNPs") +
    labs(x = "Population", y = "Test-set macro F1 (one-vs-rest)", fill = NULL) +
    theme_classic() +
    theme(axis.text.x = element_text(angle = 90, hjust = 1, vjust = 0.5)); 

print(p)
ggsave("results/fig3a_minimum_SNPs.pdf", width=21, height=9, units = c("cm"), dpi = 300)
```
